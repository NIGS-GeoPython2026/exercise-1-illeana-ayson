```python
ice_cream_rating=7
sleeping_rating=8
```

```python
first_name = input('Please enter your first name: ')
last_name = input('Please enter your last name: ')
my_name = first_name+' '+last_name
```

```python
happiness_rating = (ice_cream_rating + sleeping_rating)/2
```

```python
print('ice_cream_rating data type: ', type(ice_cream_rating),
      '\nfirst_name data type: ', type(first_name),
      '\nhappiness_rating data type: ', type(happiness_rating))
```

Yes, the data types make sense. The value stored in the ice_cream_rating is an integer. Since the first_name variable uses the input() function, it is stored as a string. As for the last variable, I thought the type() function would return 'int' since the average is a whole number, but it does make sense for the division operator to always just return a float as not all operations will return a whole number.

```python
print('My name is',first_name,'and I give eating ice cream a score of',str(ice_cream_rating), 'out of 10!')
print('I am',my_name,'and my sleeping enjoyment rating is',str(sleeping_rating),'/ 10!')
print('Based on the factors above, my happiness rating is',str(happiness_rating),'out of 10, or', str(happiness_rating*10),'%!')
```
